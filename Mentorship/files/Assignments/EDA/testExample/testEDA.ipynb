```python
import plotly.express as px

null_df = df.isnull().astype(int)
fig = px.imshow(null_df.T, aspect="auto",
                color_continuous_scale=['white','black'],
                labels=dict(x="Row", y="Column", color="Null"))
fig.update_coloraxes(showscale=False)
fig.update_layout(title="Missingness Matrix (Plotly)")
fig.show()
```

```python
import pandas as pd, numpy as np, seaborn as sns, matplotlib.pyplot as plt
import missingno as msno, plotly.express as px

df = pd.read_csv('USC00092159.csv')

# Heatmap with Seaborn
plt.figure(figsize=(12,6))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.show()

# missingno matrix
msno.matrix(df)
plt.show()

# correlation of nulls
corr = df.isnull().corr()
plt.figure(figsize=(10,8))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Null Correlation")
plt.show()

# Interactive Plotly matrix
null_df = df.isnull().astype(int)
fig = px.imshow(null_df.T, aspect="auto",
                color_continuous_scale=['white','black'],
                labels=dict(x="Row", y="Column", color="Null"))
fig.update_coloraxes(showscale=False)
fig.update_layout(title="Interactive Null Matrix")
fig.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

df = pd.read_csv('USC00092159.csv')
print(df.dtypes)

# 1. Categorical count (Matplotlib + Pandas)
df['TAVG'].value_counts().plot(kind='bar', color='skyblue')
plt.xticks(rotation=45); plt.show()

# 2. Numeric vs Category (Seaborn boxplot)
sns.boxplot(data=df, x='TAVG', y='STATION')
plt.xticks(rotation=45); plt.show()

# 3. Numeric distribution (Plotly interactive histogram)
fig = px.histogram(df, x='STATION', color='TAVG')
fig.show()
```
